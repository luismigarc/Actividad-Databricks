# Consulta de ventas desde otro notebook

En este notebook se consulta la tabla `ventas_bronze`, creada anteriormente en Unity Catalog.

El objetivo es comprobar que los datos almacenados pueden consultarse desde diferentes notebooks utilizando el nombre completo de la tabla: `catalogo.esquema.tabla`.


```python

tabla = "workspace.ingenieria_datos_luismiguel.ventas_bronze"

df_ventas = spark.table(tabla)

print("Total de registros:", df_ventas.count())
display(df_ventas)
```

## Consultas sobre las ventas

Se realizan dos consultas para verificar los datos:

1. Contar la cantidad total de ventas almacenadas.
2. Mostrar únicamente los registros correspondientes a la categoría Videojuegos.

Estas consultas permiten comprobar que la tabla Delta se puede utilizar desde otro notebook sin necesidad de crear nuevamente el DataFrame.


```python

from pyspark.sql.functions import col

print("Cantidad total de registros:", df_ventas.count())

videojuegos = df_ventas.filter(
    col("categoria") == "Videojuegos"
)

print("Ventas de videojuegos:", videojuegos.count())

display(videojuegos.orderBy("fecha"))
```

## Conclusión

Se comprobó que la tabla `ventas_bronze` puede consultarse desde un notebook diferente al utilizado para crearla.

La lectura se realizó utilizando el nombre completo de la tabla en Unity Catalog, sin necesidad de volver a cargar los datos originales.

También se verificó la cantidad total de registros y se aplicó un filtro para consultar únicamente las ventas de videojuegos.
